1. Find the largest number in the list.
numbers = [12, 45, 7, 34, 89, 23]
1.I will assume 1st number in the list as a largest
2.Then I will compare that number with other numbers in list if other number is greater than largest I will make that number as largest


```python
numbers = [12, 45, 7, 34, 89, 23] 
large = numbers[0]
for num in numbers:
  if num>large:
    large=num
print('largest',large)
```

numbers = [5, 25, 30, 10, 40, 15]
1.I will initialize count variable to 0
2.I will compare each number in list with 20
3.if num>20 I will increase the count
4. finally I will print count


```python
numbers = [5, 25, 30, 10, 40, 15]
count=0
l=[]
for num in numbers:
    if num>20:
        l.append(num)
        #print(num)
        count=count+1
print('number of elements >20',count)
print('list of elements >20',l)
```

6. Print names that start with the letter "a".
names = ["ram", "arun", "sita", "anil", "krishna"]
1.I will create an empty list names_startswithA=[]
2.I will go to each element in list and verify it start with 'a' or not.
3.if it start with a then I will append that name to my created list
4. I will print the created list

```python
names = ["ram", "arun", "sita", "anil", "krishna"] 
names_startswithA=[]
for name in names:
    if name.startswith('a'):
          names_startswithA.append(name)
print(names_startswithA)
```
