```python
# **- methods for adding elements**
# - add()
s={5,4,3}
s.add(2)
print(s)
# - update()
s.update([6,7,8])
print(s)
```

```python
# **-removing elemnts**
# - remove()--remove elemet(error if not present)
#print(s)
#s.remove(6)
# - clear()
#s.clear
# - pop()----used to remove random element in set
#s.pop()
#print(s)

# - discard()--removes element(no error if not present)
s1={88,55,25,98}
#s1.remove(6)
s1.discard(6)
```

```python
# **- mathematical set operations**
# - union()--combines sets
a={'AAA','BBB','CCC'}
b={'EEE','FFF','GGG','AAA'}
c=a.union(b)
#print(c)
# - intersection()--common elements
#print(a.intersection(b))
# - difference()--elements in first set but not in second
#print(a.difference(b))
# - symetrical difference()--elements in both but not common
print(a.symmetric_difference(b))
```
