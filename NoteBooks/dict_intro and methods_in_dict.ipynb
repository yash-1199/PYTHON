```python
#methods to access the data
#get():Returns values of a key
d={'name':'Ram','age':25,'city':'Hyd'}
print(d.get('age'))
print(d.get('name'))
print(d.get('city'))
```

```python
#keys():Return all the keys
print(d.keys())
```

```python
#values():returns all the values
print(d.values())
```

```python
#items():Returns key-value pair
d={'name':'Ram','age':25,'city':'Hyd','salary':50000,None:None,None:'abs'}
print(d.items())
```

**methods to add or update data**

```python
#update()
d1={'name':'Ram','age':25,'city':'Hyd'}
d1.update({'city':'del'})
print(d1)
```

```python
d1['gender']='male'
print(d1)
```

```python
#setdefault():returns value of key ,creates key if not present
d1.update({'city':'Wipro'})
print(d1)
```

```python
d1.setdefault('city','mum')
print(d1)
```
