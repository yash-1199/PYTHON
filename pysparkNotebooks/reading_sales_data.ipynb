```python
#to define our own schema
path='abfss://raw-sales@janbatchsa.dfs.core.windows.net/sales_60_records.csv'
from pyspark.sql.types import StructType
from pyspark.sql.types import StructField
from pyspark.sql.types import StringType,IntegerType,DateType

schema=StructType([StructField('order_id',IntegerType(),True),\
                    StructField('customer_id',StringType(),True),\
                   StructField('product_id',StringType(),True),\
                   StructField('order_date',DateType(),True),\
                   StructField('quantity',IntegerType(),True),\
                   StructField('price',IntegerType(),True)\
                  ])

df=spark.read.csv(path,schema=schema,header=True)

display(df)
```

```python
#spark API
path='abfss://raw-sales@janbatchsa.dfs.core.windows.net/sales_60_records.csv'
df=spark.read.format('csv')\
    .option('header','true')\
    .option('inferschema','true')\
    .load(path)
df.show()


```

```python
#group by customer id and lest find total amount of each customer
from pyspark.sql.functions import sum
orders=df2.groupBy('customer_id').agg(sum('total_amount')\
    .alias('total_amount'))
orders.show()
```

```python
from pyspark.sql.functions import lit,when,col
df_updated=df.withColumn('price',when(col('customer_id')=='C002',lit(None)).otherwise(col('price')))
df_updated.show()
```

```python
final_df=df_updated.na.fill(0)
final_df.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank
win=Window.partitionBy(col('customer_id')).orderBy(col('total_amount').desc())
final_df=df2.withColumn('rn',rank().over(win))
final_df.show()
```
